**Task 2.1**  
Write total_value(price, quantity) that returns the price multiplied by the quantity. Test it with the Notebook (price 1500, quantity 40).


```python
# Task 2.1
def total_value(price, quantity):
    """Return the total value of stock: price multiplied by quantity."""
    return price * quantity

print(total_value(1500, 40))   # 60000
```

**Task 2.2**  
Write apply_discount(price, percent=10) that returns the price after the discount. Call it once using the default and once with 25 percent.

```python
def apply_discount(price, percent=10):
    """Return the price after taking off the given percentage discount."""
    return price - (price * percent / 100)

print(apply_discount(1500))       # 1350.0
print(apply_discount(1500, 25))   # 1125.0
```

**Task 2.3**

Write stock_status(quantity) that returns "Out of stock" if the quantity is 0, "Low stock" if it is below 10, and "In stock" otherwise. Test it with 0, 5 and 40.

```python
def stock_status(quantity):
    """Return the stock status of the product."""
    if quantity == 0:
        return "Out of stock"
    elif quantity < 10:
        return "Low stock"
    else:
        return "In stock"
print(stock_status(0))    # Out of stock
print(stock_status(5))    # Low stock
print(stock_status(40))   # In stock
```

**Task 2.4**  
Write products_in_category(inventory, category) that takes the list of dictionaries and returns a list of the product names in that category. Test it with "Electronics".

```python
def products_in_category(inventory, category):
    """Return a list of products in the given category."""
    return [item["product"] for item in inventory if item["category"] == category]

print(products_in_category(inventory, "Electronics"))
```

This outlier pulls the mean up, so it is higher than 11 of the 14 days and doesn’t look like a normal day.
